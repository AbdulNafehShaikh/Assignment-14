```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('student_lifestyle_dataset.csv')

# Display the first 5 rows
df.head()
```

```python
# Calculate average study hours
avg_study_hours = df['study_hours'].mean()
print(f"Average Study Hours: {avg_study_hours:.2f}")
```

```python
# Group by gender and calculate average sleep hours
sleep_by_gender = df.groupby('gender')['sleep_hours'].mean()
print(sleep_by_gender.round(2))

print("\nWho sleeps more on average:", sleep_by_gender.idxmax())
```

```python
# Calculate correlation coefficient
correlation = df['screen_time'].corr(df['sleep_hours'])
print(f"Correlation between Screen Time and Sleep Hours: {correlation:.4f}")

# Optional scatter plot to visualize the relationship
plt.figure(figsize=(7, 5))
sns.scatterplot(x='screen_time', y='sleep_hours', data=df)
plt.title('Screen Time vs Sleep Hours')
plt.xlabel('Screen Time (hours)')
plt.ylabel('Sleep Hours')
plt.show()
```

```python
# Plot histogram of daily transport cost
plt.figure(figsize=(8, 5))
sns.histplot(df['daily_transport_cost'], bins=10, kde=True)
plt.title('Distribution of Daily Transport Cost')
plt.xlabel('Daily Transport Cost')
plt.ylabel('Frequency')
plt.show()
```

```python
# Group by gender and calculate average internet usage
internet_by_gender = df.groupby('gender')['internet_usage'].mean()
print(internet_by_gender.round(2))
```
