```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('pakistani_ecommerce_dataset.csv')

# Display the first 5 rows
df.head()
```

```python
# Count orders per product category
category_counts = df['product_category'].value_counts()
print(category_counts)

print("\nMost Popular Product Category:", category_counts.idxmax())
```

```python
# Calculate mean order value
avg_order_value = df['order_value'].mean()
print(f"Average Order Value: {avg_order_value:.2f}")
```

```python
# Count orders per city
city_counts = df['city'].value_counts()
print(city_counts)

print("\nCity with Highest Number of Orders:", city_counts.idxmax())
```

```python
# Count transactions per payment method
payment_counts = df['payment_method'].value_counts()
print(payment_counts)

print("\nMost Used Payment Method:", payment_counts.idxmax())
```

```python
# Group by city and calculate average order value
city_avg = df.groupby('city')['order_value'].mean().sort_values(ascending=False)
print(city_avg.round(2))

# Plot bar chart
plt.figure(figsize=(8, 5))
sns.barplot(x=city_avg.index, y=city_avg.values)
plt.title('Average Order Value by City')
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.show()
```
